```python
def get_flare(date_param, df):
    """
    Checks if there are any records in the dataframe where isFlare is True
    and largestEventDate is within the date_param and 24 hours after.

    Parameters:
    date_param (datetime): The date to check against.
    df (DataFrame): The dataframe containing the records.

    Returns:
    DataFrame: A dataframe containing the matching records.
    """
    # Convert the date_param to datetime if it is not already
    if not isinstance(date_param, pd.Timestamp):
        date_param = pd.to_datetime(date_param)

    # Define the end date (24 hours after date_param)
    end_date = date_param + timedelta(hours=24)
    end_date = pd.to_datetime(end_date).tz_localize(None)
    date_param = pd.to_datetime(date_param).tz_localize(None)

    # Filter the dataframe
    result = df[(df['Is_Flare'] == True) & 
                (pd.to_datetime(df['Largest Event Date']) >= pd.to_datetime(date_param)) & 
                (pd.to_datetime(df['Largest Event Date']) < pd.to_datetime(end_date))]

    return result
```

```python
# Find the row with the specific timestamp in the DataFrame
def get_metrics(row,plage,events,media,flare):
    try:
        guidedgradcam = retrieve_npy(str(row['local_request_date']),root='guidedgradcam',media=media)
        original = retrieve_npy(str(row['local_request_date']),root='original',media=media)
        ar_names,ar_lat,ar_lon = retrieve_ar(str(row['local_request_date']),
        plage=plage ,
        events=events,
        media=media
        )

        flare_indices = [ar_names.index(x) for x in ar_names if str(x) in flare]

        actual_pred = get_flare(row['obs_date'],flare_actual)

        pix_list = [post_hoc_explanation.convert_to_pix(float(x[1]),float(x[0])) for x in zip(ar_lat,ar_lon)]
        ars = [item for item in list(zip(ar_names,pix_list))]
        flares = [ars[x] for x in flare_indices]
        # Analyze attention map
        try:
            bounding_hulls_img_, distances_df_, score_, ratio_ = post_hoc_explanation.explain(guidedgradcam, original, flares, -20, 20, 5, 40, 30, 50, 2, 10, numpy=True)
        except Exception as e:
            print(e)
            bounding_hulls_img_, distances_df_, score_, ratio_ = None,None,None,None
        bounding_hulls_img, distances_df, score, ratio = post_hoc_explanation.explain(guidedgradcam, original, ars, -20, 20, 5, 40, 30, 50, 2, 10, numpy=True)
    except Exception as e:
        return None, None, None, actual_pred.shape[0], score_, ratio_
    return  distances_df.to_dict(orient='list'), score, ratio, actual_pred.shape[0], score_, ratio_
```

```python
db = pd.read_csv('predictions_resnet.txt')
ars = pd.read_csv('ars.csv')
flare_actual = pd.read_csv('Flares.csv')
flares = list(set(pd.read_csv(r'flares_event_ar_no.csv',dtype=str)['AR No'].to_list()))

db[['distances', 'score', 'ratio','actual', 'score_f', 'ratio_f']] = db.apply(
    lambda row: pd.Series(get_metrics(row, True, True,'resnet',flares)), axis=1   
)

# for idx,row, in db.iterrows():
#     print(idx, row['obs_date'])
#     try:
#         distances, score, ratio, actual, score_f, ratio_f,   = get_metrics(row, False, False, 'media',flares)
#     except Exception as e:
#         print(e)
```
