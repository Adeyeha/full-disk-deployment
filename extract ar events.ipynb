```python
import os
import pandas as pd

# Assuming NOAAExtractor is already defined
# from your_module import NOAAExtractor

def process_files_in_folder(folder_path):
    # Initialize an empty list to store dataframes
    dataframes = []

    # Loop through all files in the specified folder
    for filename in os.listdir(folder_path):
        if filename.endswith(".txt"):
            # Create the full file path
            file_path = os.path.join(folder_path, filename)
            # print(file_path)
            
            # Extract the dataframe from the file
            extractor = NOAAExtractor()
            try:
                df = extractor.get_noaa_dataframe(path=file_path)
            except Exception as e:
                break
            
            # Append the dataframe to the list
            dataframes.append(df)
    
    
    return dataframes


```

```python
def find_ar(location_string):
    pattern = r'\(\s*(\d+)\s*\)'

    # Find the number using the regex pattern
    match = re.search(pattern, location_string)

    if match:
        extracted_number = match.group(1)
        return extracted_number
    else:
        return None
```

```python
from datetime import datetime, timedelta
def generate_hourly_timestamps(start_timestamp, end_timestamp):
    # Convert string timestamps to datetime objects
    # start = datetime.strptime(start_timestamp, '%Y-%m-%d %H:%M:%S')
    # end = datetime.strptime(end_timestamp, '%Y-%m-%d %H:%M:%S')

    start = start_timestamp
    end = end_timestamp
    # Initialize list to store timestamps
    hourly_timestamps = []

    # Generate timestamps with 1-hour interval
    current = start
    while current <= end:
        hourly_timestamps.append(current)
        current += timedelta(hours=4)

    return hourly_timestamps
```

```python
import pandas as pd
import ast
import re

dd=pd.read_csv(r'lmsal_events_with_linked_data_new.csv')
```

```python
dd['Linked Data'][0]
```

```python
def extract_fields(cell):
    events_list = ast.literal_eval(cell)
    extracted_data = []
    for event in events_list:
        derived_position = event['Derived Position']
        ar_no_match = re.search(r'\(\s*(\d+)\s*\)', derived_position) if derived_position else None
        ar_no = ar_no_match.group(1) if ar_no_match else None
        class_ = event['GOES Class'][0]
        
        extracted_data.append({
            'EName': event['EName'],
            'Start': event['Start'],
            'Stop': event['Stop'],
            'Peak': event['Peak'],
            'GOES Class': event['GOES Class'],
            'Class': class_,
            'Derived Position': derived_position,
            'AR No': ar_no
        })
    return extracted_data

# Apply the extraction function to the DataFrame column and flatten the resulting lists into a DataFrame
extracted_rows = dd['Linked Data'].apply(extract_fields).explode().reset_index(drop=True)
result_df = pd.json_normalize(extracted_rows)
result_df['AR No'] = '1' + result_df['AR No']
```
